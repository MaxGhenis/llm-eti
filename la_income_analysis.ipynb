# Los Angeles Income Percentile Analysis

This notebook calculates the 85th percentile household income for the City of Los Angeles using American Community Survey (ACS) data.

```python
import censusdata
import numpy as np
import pandas as pd
```

## Fetch ACS Data for Los Angeles City

We use the ACS 2023 5-year estimates (2019-2023), table B19001 (Household Income in the Past 12 Months).

Geographic identifiers:
- State: 06 (California)
- Place: 44000 (Los Angeles city)

```python
# Fetch data for Los Angeles city
la_city = censusdata.download(
    "acs5",
    2023,
    censusdata.censusgeo([("state", "06"), ("place", "44000")]),
    ["B19001_001E"]  # Total households
    + [f"B19001_{str(i).zfill(3)}E" for i in range(2, 18)],  # Income brackets
)

print(f"Data retrieved for {len(la_city)} geography")
la_city
```

## Define Income Brackets

ACS provides income data in brackets. We use the midpoint of each bracket for analysis.

```python
income_brackets = {
    "B19001_002E": 5000,      # < $10k
    "B19001_003E": 12500,     # $10k-$14,999
    "B19001_004E": 17500,     # $15k-$19,999
    "B19001_005E": 22500,     # $20k-$24,999
    "B19001_006E": 27500,     # $25k-$29,999
    "B19001_007E": 32500,     # $30k-$34,999
    "B19001_008E": 37500,     # $35k-$39,999
    "B19001_009E": 42500,     # $40k-$44,999
    "B19001_010E": 47500,     # $45k-$49,999
    "B19001_011E": 55000,     # $50k-$59,999
    "B19001_012E": 67500,     # $60k-$74,999
    "B19001_013E": 87500,     # $75k-$99,999
    "B19001_014E": 112500,    # $100k-$124,999
    "B19001_015E": 137500,    # $125k-$149,999
    "B19001_016E": 175000,    # $150k-$199,999
    "B19001_017E": 250000,    # $200k+
}
```

## Calculate Income Distribution

```python
# Create distribution weighted by household counts
income_distribution = []

for var, income in income_brackets.items():
    count = la_city.iloc[0][var]
    income_distribution.extend([income] * int(count))

print(f"Total households in distribution: {len(income_distribution):,}")
```

## Results: Income Percentiles

```python
# Calculate key percentiles
percentiles = [25, 50, 75, 85, 90, 95]
results = {}

for p in percentiles:
    results[f"{p}th percentile"] = np.percentile(income_distribution, p)

# Display results
print("="*60)
print("Los Angeles City Household Income Percentiles")
print("ACS 2023 5-Year Estimates (2019-2023)")
print("="*60)
print(f"Total Households: {la_city.iloc[0]['B19001_001E']:,}")
print()
for label, value in results.items():
    print(f"{label:20s}: ${value:,.0f}")
print("="*60)
```

## Summary

**The 85th percentile household income in Los Angeles city is calculated from the latest ACS data.**

This means that 85% of households in Los Angeles earn less than this amount, based on ACS 2023 5-year estimates (2019-2023).
